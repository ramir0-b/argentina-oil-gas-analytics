# Data Preparation

## Objectives

- Remove administrative metadata not relevant for analytics
- Convert date columns to datetime format
- Preserve analytical attributes for future reporting and modeling
- Create a cleaned dataset for downstream consumption

## Transformations Applied

### Removed Columns

- habilitado
- rectificado
- idusuario
- observaciones

### Date Conversion

Converted:

- fechaingreso
- fecha_data

to datetime format.

### Output Dataset

The cleaned dataset was exported to:

data/processed/argentina_oil_gas_clean.csv

```python
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)

df = pd.read_csv('../data/raw/produccin-de-pozos-de-gas-y-petrleo-2026.csv')
```

```python
df_clean = df.copy()
```

```python
df.columns
```

```python
selected_columns = ['idempresa', 'anio', 'mes', 'idpozo', 'prod_pet', 'prod_gas',
       'prod_agua', 'iny_agua', 'iny_gas', 'iny_co2', 'iny_otro', 'tef',
       'vida_util', 'tipoextraccion', 'tipoestado', 'tipopozo','fechaingreso',
        'empresa', 'sigla', 'formprod', 'profundidad', 'formacion',
       'idareapermisoconcesion', 'areapermisoconcesion', 'idareayacimiento',
       'areayacimiento', 'cuenca', 'provincia', 'tipo_de_recurso', 'proyecto',
       'clasificacion', 'subclasificacion', 'sub_tipo_recurso', 'fecha_data']
```

Omitting the following columns:
* habilitado
* rectificado
* idusuario
* observaciones

```python
df.head()
```

```python
df.tail()
```

The dataset already contains a monthly reference date (fecha_data). Therefore, no additional date column was created.

```python
df_clean = df_clean[selected_columns]
```

```python
print(df.shape)
print(df_clean.shape)
```

```python
df_clean.info()
```

```python
df_clean["fechaingreso"] = pd.to_datetime(df_clean["fechaingreso"])
df_clean["fecha_data"] = pd.to_datetime(df_clean["fecha_data"])
```

```python
df_clean.info()
```
